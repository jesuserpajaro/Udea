### **1.4. Relación Entre Variables**
---

- Analice si existe alguna relación entre las variables. Puede utilizar medidas de correlación o visualizaciones.
- Identifique si hay redundancia entre las variables para una posterior eliminación o selección.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

#Columnas para la correlación
cols_to_include = ['Marital status', 'GDP', 'Previous qualification']

# dataframe con las columnas seleccionadas
subset = data[cols_to_include].copy()

# Convertir columnas categóricas a numéricas
for col in ['Marital status', 'Previous qualification']:
    if subset[col].dtype == 'object' or str(subset[col].dtype).startswith('category'):
        le = LabelEncoder()
        subset[col] = le.fit_transform(subset[col])

# Calcular la matriz de correlación
corr = subset.corr()

# Graficar
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Matriz de correlación")
plt.show()
```

```python
# Muestra algunas variables numéricas para ver relaciones
sns.pairplot(data.select_dtypes(include='number').sample(5, axis=1))
plt.show()
```

```python
# Revisar redundancia por alta correlación
threshold = 0.9
upper = corr.where(~np.tril(np.ones(corr.shape)).astype(bool))
redundant_features = [column for column in upper.columns if any(upper[column].abs() > threshold)]

print("Variables altamente correlacionadas (redundantes):")
print(redundant_features)
```

### **1.3. Valores atípicos**
---
Al abordar valores atípicos, es relevante cuestionarse sobre la naturaleza de estos puntos extremos.
* ¿Son errores de medición o representan información válida pero excepcional?

Los valores atípicos no deben considerarse simplemente como errores de medición, ya que contienen información significativa en cualquier análisis estadístico. Estos valores extremos nos permiten comprender el comportamiento de una minoría dentro de la muestra o población. Ignorarlos podría llevar, dependiendo del tipo de análisis, a que el modelo no considere escenarios poco probables pero altamente relevantes, como aquellos asociados a pérdidas catastróficas. Un ejemplo de esto es la pérdida máxima esperada por un siniestro, como un terremoto, en una aseguradora, lo cual podría afectar gravemente el patrimonio de la compañía.


La elección entre eliminar o transformar estos valores debe basarse en su influencia en la distribución y la interpretación de los resultados, asegurando que el análisis refleje de manera precisa la realidad del conjunto de datos.


* ¿Qué criterios o técnicas se utilizaron para identificar los valores atípicos?


Dado que las variables continuas también estaban delimitadas por intervalos, técnicas como los percentiles no fueron suficientemente eficaces para identificar valores atípicos. Sin embargo, considerando que la mayoría de las variables eran discretas, optamos por un método basado en el conteo. Este consistió en revisar, para cada variable, cuáles valores presentaban una baja frecuencia de aparición. Un valor con una muestra poco representativa dentro de una columna puede introducir ruido al análisis, ya que el modelo podría aprender un patrón erróneo al asociar dicha característica con un comportamiento específico. A continuación, se presenta un ejemplo de cómo utilizar porcentajes para identificar cuándo un valor en una columna no es estadísticamente significativo:

```python
Datos = data.copy()
Columna = 'Marital status'
Agrupacion = Datos.groupby([Columna]).agg(Conteo=(Columna, 'count')).reset_index().sort_values(by='Conteo',ascending=False)
Agrupacion['Porcentaje'] = (Agrupacion['Conteo'] / Agrupacion['Conteo'].sum())*100
Agrupacion['Graficar'] = Agrupacion.apply(lambda df: '' if df['Porcentaje'] < 5 else df[Columna], axis=1)
Agrupacion
```

* ¿Se aplicaron métodos estadísticos o visuales para detectar los valores atípicos?
Sí, se aplicó estadística descriptiva para analizar las proporciones mediante conteos y, con base en el criterio del analista, establecer un porcentaje mínimo que deben cumplir las variables para que un registro sea considerado. Además, se elaboraron gráficos que permiten visualizar estas proporciones de forma más clara, como se muestra a continuación:

```python
import matplotlib.pyplot as plt

# Función para mostrar porcentaje solo si es >= 5%
def mostrar_pct(pct):
    return f'{pct:.1f}%' if pct >= 5 else ''

# Paleta de colores pastel personalizada
colores = ['#FF9999', '#66B3FF', '#99FF99', '#FFCC99', '#C2C2F0', '#FFB3E6', '#FFD700', '#B0E0E6']

plt.figure(figsize=(6, 6))
plt.pie(
    Agrupacion['Porcentaje'],
    labels=Agrupacion['Graficar'],
    autopct=mostrar_pct,
    startangle=90,
    labeldistance=1.1,
    pctdistance=0.85,
    colors=colores,  # Se aplica la paleta aquí
    wedgeprops={'linewidth': 1, 'edgecolor': 'white'}
)
plt.title(f'Distribución de {Columna}', fontsize=16, fontweight='bold', pad=30)
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
Agrupacion['Porcentaje'].plot(kind='bar', color='m')
plt.title('Grafico de barras')
plt.xlabel(Columna)
plt.ylabel('Porcentaje en cientos')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

#Cargar los datos
url = 'https://raw.githubusercontent.com/jesuserpajaro/diplomado-/refs/heads/main/data.csv'
data = pd.read_csv(url, sep=';')
X = data.drop('Target', axis=1)
y = data['Target']
X_encoded = pd.get_dummies(X)

#Separar entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.3, random_state=42, stratify=y)

#Búsqueda de hiperparámetros
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [5, 10, 20, None],
    'min_samples_split': [2, 5, 10]
}
grid_search = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5, scoring='f1_weighted', n_jobs=-1)
grid_search.fit(X_train, y_train)

#Resultados del mejor modelo
best_model = grid_search.best_estimator_
print("=== Mejor combinación de hiperparámetros ===")
print(grid_search.best_params_)

y_pred_best = best_model.predict(X_test)
print(classification_report(y_test, y_pred_best))
ConfusionMatrixDisplay.from_estimator(best_model, X_test, y_test, cmap="Greens")
plt.title("Matriz de Confusión - Modelo Optimizado")
plt.show()

#Curvas de desempeño
results = pd.DataFrame(grid_search.cv_results_)

# Gráfica: efecto de n_estimators sobre f1_score
pivot_estimators = results.pivot_table(index='param_n_estimators', values='mean_test_score')
pivot_estimators.plot(marker='o')
plt.ylabel('F1 Score (prom. validación)')
plt.title('Desempeño vs n_estimators')
plt.grid(True)
plt.show()

# Gráfica: efecto de max_depth
pivot_depth = results.pivot_table(index='param_max_depth', values='mean_test_score')
pivot_depth.plot(marker='s', color='orange')
plt.ylabel('F1 Score (prom. validación)')
plt.title('Desempeño vs max_depth')
plt.grid(True)
plt.show()
```
